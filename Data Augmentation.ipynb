 data augmentation

Data Augmentation is a machine learning technique that artificially expands the size and diversity of a training dataset without collecting new data. It applies randomized transformations to existing samples—such as rotating, cropping, or adding noise—creating modified copies while preserving the original labels.
Core Objectives

    Prevents Overfitting: Prevents the model from memorizing specific training samples and their exact pixel/feature orientations.

    Improves Generalization: Teaches the model features that are invariant to scaling, orientation, lighting, or background noise.

    Balances Imbalanced Data: Generates additional synthetic samples for minority or underrepresented classes.

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
import matplotlib.pyplot as plt

# Load image
img = load_img('Sample_img.jpg')   # your image path

```

```python
x = img_to_array(img)
```

```python
x.shape
```

```python
x = x.reshape((1,) + x.shape)
```

```python
x.shape
```

```python
# Define augmentation
datagen = ImageDataGenerator(
    rotation_range=40,
    zoom_range=0.3,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Generate and display images
i = 0
plt.figure(figsize=(10,5))

for batch in datagen.flow(x, batch_size=1):
    plt.subplot(3, 3, i+1)
    plt.imshow(batch[0].astype('uint8'))
    plt.axis('off')
    
    i += 1
    if i == 9:   # show 9 images
        break

plt.show()
```
